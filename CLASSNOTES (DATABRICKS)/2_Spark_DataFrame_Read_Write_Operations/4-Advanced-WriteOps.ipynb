###We are going to learn some advanced options available in the different file/table write functions (Least bother)

#### Important - There are 4 typical write modes and default is error.
1. overwrite
2. append
3. ignore
4. error


### CSV Write Advanced Feature - Very Important, Important, Not Important (just try to know once for all)


**Very Important:** 

path, mode, compression, sep, quote, escape, header 

**Important:** 

nullValue,quoteAll,dateFormat, timestampFormat, ignoreLeadingWhiteSpace, ignoreTrailingWhiteSpace,lineSep 

**Not Important:** 

escapeQuotes, charToEscapeQuoteEscaping, encoding, emptyValue

```python
struct1="custid int,name string,age int"

df1=spark.read.schema(struct1).csv("/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/malformeddata1.txt",
                                    header=False,
                                    sep=',',
                                    mode='permissive',
                                    comment='#',
                                    quote="'",
                                    escape="|"
                                )
df1.show(10,False)
```

```python
df1.write.csv( path="/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/csvadvwrite/", 
                mode='overwrite', 
                compression='None', 
                sep=',', 
                quote="'", 
                escape='~', 
                header=True
            )
```

```python
df1.write.csv( path="/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/csvadvwrite/", 
                mode='overwrite', 
                compression='None', 
                sep=',', 
                quote="'", 
                escape='~', 
                header=True,
                nullValue='na',
                quoteAll=True,
                lineSep='-'
            )
#nullValue,quoteAll,dateFormat, timestampFormat, ignoreLeadingWhiteSpace, ignoreTrailingWhiteSpace,lineSep
```


##JSON Write Advanced Feature - Very Important, Important, Not Important (just try to know once for all)


**Very Important:** 

path, mode, compression

**Important:** 

dateFormat, timestampFormat, lineSep, ignoreNullFields

**Not Important:**

```python
df1.write.json(path="/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/jsonadvwrite/", 
                mode="append", 
                compression="None",
                ignoreNullFields=True
            )
```


### Serialized Data format (orc/parquet/delta(we learn indepth in project))


**Very Important:** 

path, mode, partitionBy, compression 

```python
df1.write.orc(path="/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/orcadvwrite/",
                mode='error',
                compression='snappy',
                partitionBy='age'
            )
```

```python
df1.write.parquet(path="/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/parquetadvwrite/",
                  mode="error",
                  compression='snappy',
                  partitionBy='age'
                )
```

```python
spark.read.orc("/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/orcadvwrite/") \
          .where("age=30") \
          .display()
```

```python
spark.read.parquet("/Volumes/catalog1_dropme/schema1_dropme/volume1_dropme/dir1_dropme/data/ingestion_volume/target/parquetadvwrite/") \
         .where("age=30") \
         .display()
```
