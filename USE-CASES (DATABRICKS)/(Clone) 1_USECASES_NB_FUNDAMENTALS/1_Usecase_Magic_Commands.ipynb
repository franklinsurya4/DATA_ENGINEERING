
## Task-3: 

### Create a child notebook "4_child_nb_dataload" and just do a python print of your name.

**Later:** 

- write code to load data, Using the requests library, 
- perform api call to pull data from "https://public.tableau.com/app/sample-data/mobile_os_usage.csv" into a python variable using the magic command %py and 
- write the data into the created volume "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv" using the above variable.
- text using the magic command dbutils.fs.put("volume",variable.text,overwrite=True)

```python
%python

import requests

# API URL
url = "https://public.tableau.com/app/sample-data/mobile_os_usage.csv"

# API call
response = requests.get(url)

# Validate API response
response.raise_for_status()

# Store CSV content in a Python variable
variable = response

print("API Status Code:", response.status_code)

print("Data Loaded Successfully")

print(variable.text[:500])
```


## Now, write the downloaded CSV content into the volume.

```python
%python

# Volume file path
file_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

# Write API response text into the volume
dbutils.fs.put(
    file_path,
    variable.text,
    overwrite=True
)

print("CSV file successfully written into the volume")
```


## Task-6:

### Create a pyspark dataframe df1 reading the data from the above file using pyspark magic command %python


### Read the CSV file into a PySpark DataFrame.

```python
%python

file_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

df1 = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
)

# Display DataFrame
display(df1)
```


## Check the schema:

```python
%python

df1.printSchema()
```


### Check the total number of records:

```python
%python

print("Total Records:", df1.count())
```


## Task-7: 

### Write the above dataframe df1 data into a databricks table called 'default.mobile_os_usage' using pyspark magic command %python

```python
%python

df1.write \
    .format("delta") \
    .option("delta.columnMapping.mode", "name") \
    .option("delta.minReaderVersion", "2") \
    .option("delta.minWriterVersion", "5") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.mobile_os_usage")
```


## Task-8: 

### Write sql query to display the data loaded into the table 'default.mobile_os_usage' using the pyspark magic command %python 

```python
%python

result_df = spark.sql("""
    SELECT *
    FROM workspace.default.mobile_os_usage
""")

display(result_df)
```
