```python
%python

# Base Volume Path
base_path = "/Volumes/workspace/default/volumewd36"

# Define folder paths
raw_path = f"{base_path}/raw"

processed_path = f"{base_path}/processed"

archive_path = f"{base_path}/archive"

# Display folder paths
print("HEALTHCARE DATA STORAGE PATHS")
print("=" * 60)

print("BASE PATH      :", base_path)
print("RAW PATH       :", raw_path)
print("PROCESSED PATH :", processed_path)
print("ARCHIVE PATH   :", archive_path)
```


## 2. dbutils Usecase:


### Create Directories using the above path variables..

```python
%python

# Create directories

dbutils.fs.mkdirs(raw_path)

dbutils.fs.mkdirs(processed_path)

dbutils.fs.mkdirs(archive_path)

print("Healthcare directories created successfully!")
```


## 3. dbutils Usecase:

**sample_data = """patient_id,patient_name,age,gender

```python
%python

# Sample healthcare data

sample_data = """patient_id,patient_name,age,gender
1,John Doe,68,M
2,Jane Smith,54,F
"""

# Define file path
raw_file_path = f"{raw_path}/healthcare_patients.csv"

# Write data into raw folder
dbutils.fs.put(
    raw_file_path,
    sample_data,
    overwrite=True
)

print("Healthcare sample file created successfully!")
```


## 4. dbutils Usecase:


list the file created


**TODO:**

```python
%python

# List files in raw folder

raw_files = dbutils.fs.ls(raw_path)

print("FILES AVAILABLE IN RAW FOLDER")
print("=" * 60)

for file in raw_files:
    print(file.name, file.path, file.size)
```


## 5. dbutils Usecase:


### Copy File (raw → processed)

```python
%python

# Source file
raw_file_path = f"{raw_path}/healthcare_patients.csv"

# Destination file
processed_file_path = f"{processed_path}/healthcare_patients.csv"

# Copy file
dbutils.fs.cp(
    raw_file_path,
    processed_file_path
)

print("Healthcare file copied successfully!")
print("Source      :", raw_file_path)
print("Destination :", processed_file_path)
```


**dbutils.fs.cp()** is used to `copies a file from one storage location to another`.

## 6. dbutils widget usecase:



 - Create **dropdown** and **text** widgets...


**TODO:** 

  - Create a **dropdown widget** for `environment (dev, qa, prod)`

**TODO:** 

  - Create a **text widget** for `owner name`

```python
%python

# Create dropdown widget
dbutils.widgets.dropdown(
    "environment",
    "dev",
    ["dev", "qa", "prod"],
    "Select Environment"
)

# Create text widget
dbutils.widgets.text(
    "owner",
    "FRANKLIN SURYA S",
    "Owner Name"
)

print("Widgets created successfully!")
```


## 7. dbutils widget Usecase:


### Read Widget Values environment and owner and print in the screen

```python
%python

# Read widget values

environment = dbutils.widgets.get("environment")

owner = dbutils.widgets.get("owner")

# Display values

print("NOTEBOOK CONFIGURATION")
print("=" * 50)

print("Environment :", environment)

print("Owner       :", owner)
```


## 8. dbutils widget Usecase:


 Move the above **processed File to Archive**.

```python
%python

# Source file
processed_file_path = f"{processed_path}/healthcare_patients.csv"

# Destination file
archive_file_path = f"{archive_path}/healthcare_patients.csv"

# Move file
dbutils.fs.mv(
    processed_file_path,
    archive_file_path
)

print("Healthcare file moved successfully!")
print("Source      :", processed_file_path)
print("Destination :", archive_file_path)
```


**dbutils.fs.mv()** moves a `file from one location to another`.
