## Q24. What are the types of Databricks Volumes?


Databricks supports **two types** of Unity Catalog Volumes:


**1. Managed Volume:**


- Databricks manages storage location.
- Unity Catalog manages permissions.
- Suitable for files whose storage location is managed through Databricks.


**2. External Volume:**


- References an existing external cloud storage location.
- Unity Catalog governs access to the files.
- Suitable when organizations already maintain cloud storage locations.


| Managed Volume | External Volume |
|---|---|
| Managed storage location | Existing external storage location |
| Databricks manages storage placement | References external storage |
| Unity Catalog permissions | Unity Catalog permissions |

```python
CREATE VOLUME IF NOT EXISTS

workspace.default.data_volume;


### Step 2: List files


```python
display(dbutils.fs.ls("/Volumes/workspace/default/data_volume/"))


### Step 3: Read CSV


```python
df = (spark.read.option("header", True)    
.option("inferSchema", True)    
.csv("/Volumes/workspace/default/data_volume/customers.csv"))
display(df)


### Step 4: Check records


```python
print("Total records:", df.count())df.printSchema()


## Q26. What is the difference between DBFS and Unity Catalog Volumes?



- **DBFS** stands for `Databricks File System`. 
- It is a `legacy filesystem abstraction` used in **older Databricks environments**.
- **Unity Catalog Volumes** are `governed file-storage objects` integrated with `Unity Catalog`.


| DBFS | Unity Catalog Volumes |
|---|---|
| Legacy filesystem abstraction | Unity Catalog-governed file storage |
| Legacy DBFS root may be disabled | Recommended for governed file access |
| Legacy paths and access patterns | Supports Unity Catalog permissions |
| Often found in older projects | Suitable for modern governed architectures |


## SECTION 7: DATABRICKS GIT FOLDERS (REPOS)


## Q27. What are Databricks Repos?


- **Databricks Repos**, now commonly called `Git folders`.
- It provides `Git integration` inside the `Databricks workspace`.
- They allow developers to manage `notebooks` and `project files` using **version control systems** such as `GitHub, GitLab, and Azure DevOps.`


| Operation | Purpose |
|---|---|
| Clone | Create a workspace copy of a remote repository |
| Commit | Save changes in local Git history |
| Push | Upload committed changes to remote repository |
| Pull | Download and integrate remote changes |
| Branch | Create an independent development line |
| Merge | Combine changes from different branches |



## Q28. How do you connect GitHub to Databricks?


**Follow these steps:**

1. Open Databricks Workspace.
2. Navigate to Git folders.
3. Click Create Git folder.
4. Select GitHub as the Git provider.
5. Authenticate using configured Git credentials.
6. Enter the GitHub repository URL.
7. Select the required branch.
8. Click Create.
Example repository:
https://github.com/franklinsurya4/Data_Engineering_Python


## Q29. Explain the Git folder workflow.


The **Git folder workflow** means `developers to maintain and synchronize project code` between **Databricks** and **GitHub**.


**Workflow:**


```text
GitHub Repository
       |
       v
Clone Repository
       |
       v
Databricks Git Folder
       |
       v
Create / Modify Notebook
       |
       v
Review Changes
       |
       v
Commit Changes
       |
       v
Push Changes
       |
       v
GitHub Updated


### Step 1: Clone


Connect the `remote GitHub repository` to Databricks.


### Step 2: Develop


`Create or modify notebooks` inside the Git folder.


### Step 3: Commit


`Save the changes` with a **meaningful commit message**.


**Example:**

Added reusable Bronze ingestion notebook


### Step 4: Push


`Upload committed changes` to GitHub.


### Step 5: Pull


`Download changes` made by other developers.


### Step 6: Merge


Merge approved changes into the main branch using the team's Git workflow.


## Q30. What is the difference between Workspace notebooks and Git folder notebooks?


- **Workspace notebooks** are maintained as `Databricks workspace artifacts`. 
- **Git folder notebooks** are maintained inside `Git-integrated folders`.




| Workspace Notebooks | Git Folder Notebooks |
|---|---|
| Stored in Databricks Workspace | Stored in Git-backed folders |
| Not automatically version-controlled by Git | Git version control supported |
| Useful for workspace development | Useful for collaborative development |
| Changes are not automatically pushed to GitHub | Changes can be committed and pushed |
| Workspace permissions apply | Workspace and Git permissions apply |


## SECTION 8: DAY-TO-DAY DATABRICKS USAGE


## Q31. Explain your daily activities as a Databricks Data Engineer.


In my day-to-day work, I use Databricks to develop, execute, monitor, and maintain data engineering pipelines.


**My typical activities include:**

- Opening the Databricks Workspace and accessing project notebooks.
- Connecting to the required compute environment.
- Reading incoming CSV or Parquet files from Unity Catalog Volumes.
- Using PySpark to perform data cleaning and transformations.
- Writing processed data into Bronze, Silver, and Gold Delta tables.
- Using widgets to parameterize notebooks.
- Retrieving credentials using Databricks Secrets.
- Using Git folders to commit and push notebook changes.
- Configuring Databricks Workflows for automated execution.
- Monitoring job execution and troubleshooting failures.
- Validating output tables using Spark SQL.
- Supporting dashboard requirements through Gold-layer datasets.


## Q32. Explain how you set up and parameterize a notebook.


**Notebook parameterization** allows us to `pass multiple values` into a `notebook` without `changing its source code`.


- Suppose we have seven E-Commerce source datasets. 
- Instead of creating seven ingestion notebooks, I create one reusable notebook.


### Step 1: Create widgets


```python
dbutils.widgets.text("source_file","orders.csv")
dbutils.widgets.text("target_table","orders")
dbutils.widgets.dropdown("environment","dev",   
["dev", "test", "prod"])


### Step 2: Retrieve parameters


```python
source_file = dbutils.widgets.get("source_file")
target_table = dbutils.widgets.get("target_table")
environment = dbutils.widgets.get("environment")


### Step 3: Configure source path


```python
BASE_PATH = ("/Volumes/workspace/default/data_volume/ecommerce_lakehouse")
source_path = f"{BASE_PATH}/landing/{source_file}"


### Step 4: Read source data


```python
df = (spark.read.option("header", True)    
                .option("inferSchema", True)    
                .csv(source_path))


## Q33. How do you connect a Git folder and push a notebook?


First, I connect my GitHub repository to Databricks using Git folders.


**Then I follow these steps:**

- Create or open a Git folder.
- Select the required branch.
- Create or modify the notebook.
- Review the changed files.
- Enter a meaningful commit message.
- Commit the changes.
- Push the committed changes to GitHub.
- Verify the updated notebook in the remote repository.


**Example commit message:**

Added parameterized Bronze ingestion notebook.


## SECTION 9: SCENARIO-BASED INTERVIEW QUESTIONS


## Q34. Your notebook works on All-purpose compute but fails on Serverless compute. What will you check?


I first check whether the notebook uses any `APIs, libraries, configurations, or filesystem operations` that are `unsupported in the serverless environment`.


**I investigate:**

- Unsupported Spark APIs.
- RDD-related operations.
- Restricted filesystem access.
- Unity Catalog permissions.
- Python package compatibility.
- Network connectivity.
- Unsupported Spark configurations.


**Interview tip:**

Never assume every classic-cluster feature works identically in serverless compute.


## Q35. How do you prevent passwords from appearing in notebooks?


I use Databricks Secrets to securely store sensitive credentials instead of hardcoding passwords inside notebooks.


**My approach:**

1. Create a secret scope.
2. Store credentials as secrets.
3. Configure appropriate ACL permissions.
4. Retrieve credentials using dbutils.secrets.get().
5. Pass credentials to the JDBC connection.
6. Avoid printing credentials in notebook outputs or logs.


**Code:**


```python
password = dbutils.secrets.get(scope="ecommerce-scope",  key="db-password")


## Q36. Your notebook needs to process different tables every day. How do you handle it?


I use notebook parameterization with Databricks Widgets and pass the required table name through Databricks Workflows.


**Example:**


```python
dbutils.widgets.text("table_name", "orders")
table_name = dbutils.widgets.get("table_name")
source_table = f"workspace.ecommerce_bronze.{table_name}"
df = spark.table(source_table)display(df)


## Q37. How do you share notebooks between developers?


I use Databricks Git folders integrated with GitHub.


**My workflow is:**

- Connect the repository.
- Create a feature branch.
- Develop notebooks.
- Commit the changes.
- Push changes to the remote repository.
- Create a pull request.
- Review and merge approved changes.


## Q38. How do you manage CSV files in Databricks?


I use Unity Catalog Volumes to store `incoming CSV files` and `Spark` or `Auto Loader` to process them.


**My typical workflow:**

1. Upload or receive CSV files in the landing location.
2. Read the files using Spark or Auto Loader.
3. Validate the schema.
4. Check missing and invalid records.
5. Add ingestion metadata.
6. Write the records into Bronze Delta tables.
7. Apply transformations in Silver.


**Example:**


```python
df = (spark.read.option("header", True)    
.option("inferSchema", True)    
.csv("/Volumes/workspace/default/data_volume/orders.csv"))

display(df)



## Q39. A developer accidentally deletes a file from a Volume. How would you recover it?


- First, I check whether the original source file is available in the upstream system, backup location, or archive.
- If the file is unavailable, I check whether cloud-storage versioning, backup, or recovery mechanisms were configured.
- If a recoverable version exists, I restore it to the Volume and verify its contents.


## Q40. What happens when a Databricks cluster terminates?


When a Databricks cluster terminates, its compute resources are released and Spark processing stops.

**The following things happen:**

1. Running Spark jobs stop or fail.
2. In-memory cached DataFrames are lost.
3. Driver and executor processes terminate.
4. Compute resources are released.
5. Workspace notebooks remain available.
6. Persistent Delta tables remain stored in cloud storage.
7. Data stored in Unity Catalog Volumes remains available.
