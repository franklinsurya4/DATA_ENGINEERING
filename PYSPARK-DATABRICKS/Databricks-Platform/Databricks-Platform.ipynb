## Q4. What is a Databricks cluster?


A **Databricks cluster** is a `collection of compute resources` used to `execute distributed data-processing workloads` using **Apache Spark**.


**It generally consists of:**

**1. Driver Node**

- Coordinates Spark execution.
- Creates the Spark execution plan.
- Schedules tasks.
- Collects task results.

**2. Worker Nodes**

- Execute distributed tasks.
- Process data partitions.
- Perform transformations.
- Store intermediate computation data.

**3. Executors**

- Execute tasks assigned by the driver.
- Perform computations.
- Maintain intermediate data and cached partitions.

```python
from pyspark.sql import functions as F

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(
        "/Volumes/workspace/default/data_volume/employees.csv"
    )
)

display(df)


**Real-time example:**

I create a notebook called 01_bronze_orders_ingestion to read incoming order CSV files and ingest them into the Bronze Delta table.


## Q10. How do you import and export notebooks?


Databricks allows us to `import and export notebooks` in supported formats.


**Import process:**

1. Open Workspace.
2. Select the destination folder.
3. Click Import.
4. Upload the exported notebook or supported source file.
5. Open and execute the notebook.



**Export process:**

1. Open the notebook.
2. Select the notebook menu.
3. Choose Export.
4. Select the required format.
5. Download the exported notebook.


**Common formats:**

- SOURCE
- IPYNB
- HTML
- DBC


**Real-time example:**

I can export a notebook as an IPYNB file for local development or import a notebook from another environment into my Databricks workspace.


## Q11. What are Databricks magic commands?


- **Magic commands** are `special notebook commands` used to **execute different operations**, 
- such as `Python code, SQL queries, shell commands, filesystem operations, and notebook execution`.



| Command | Purpose |
|---|---|
| `%python` | Execute Python code |
| `%sql` | Execute SQL queries |
| `%md` | Write Markdown documentation |
| `%sh` | Execute shell commands |
| `%fs` | Perform filesystem operations |
| `%run` | Execute another notebook |


**Practical examples:**


**1. Python**

%pythonprint("E-Commerce Lakehouse Project")

**2. SQL**

%sql
SELECT *
FROM workspace.ecommerce_bronze.orders
LIMIT 10;

**3. Markdown**

%md

**4. Shell**

%sh
pwd

**5. Filesystem**

%fs ls /Volumes/workspace/default/data_volume/

**6. Run another notebook**

%run ./common_functions


**Real-time example:**

I use %python for PySpark transformations, %sql to validate Delta tables, %md to document pipeline steps, and %run to reuse common transformation functions.


## Q12. What is the difference between %run and dbutils.notebook.run()?


| %run | dbutils.notebook.run() |
|---|---|
| Executes notebook in current context | Executes notebook as a separate notebook run |
| Shares variables and function definitions | Separate execution context |
| Useful for reusable code | Useful for notebook orchestration |
| Does not provide timeout argument | Supports timeout |
| No exit-value string return | Returns exit value as a string |


**Example using %run:**

- %run ./common_functions

**If common_functions defines:**

- def clean_data(df):    return df.dropDuplicates()

**We can call it in the current notebook:**

- cleaned_df = clean_data(df)

**Example using dbutils.notebook.run():**

result = dbutils.notebook.run(    "./silver_transformation",    600,    {        "table_name": "orders"    })print(result)


**Real-time example:**

I use %run to reuse common Python functions and dbutils.notebook.run() when I need to execute another notebook with parameters and retrieve its exit value.


**Interview tip:**

For complex production orchestration, Databricks Workflows with separate tasks and dependencies are generally preferable to deeply nested notebook calls.


## SECTION 3: DATABRICKS UTILITIES – DBUTILS


## Q13. What is dbutils?


- **dbutils** stands for `Databricks Utilities`.
- It provides `utility functions` to `interact with the Databricks environment`.



**Important modules:**

| Module | Purpose |
|---|---|
| `dbutils.fs` | File operations |
| `dbutils.secrets` | Retrieve secrets |
| `dbutils.widgets` | Create notebook parameters |
| `dbutils.notebook` | Execute and exit notebooks |
| `dbutils.library` | Library-related operations where supported |


**Real-time example:**

In my E-Commerce project, I use dbutils.fs to inspect landing files, dbutils.widgets to pass table names, and dbutils.notebook to execute reusable notebooks.


## Q14. How do you use dbutils.fs?


**dbutils.fs** is used to` perform filesystem operations on supported storage locations`.

**1. List files:**

- dbutils.fs.ls(    "/Volumes/workspace/default/data_volume/")

**2. Create directory:**

- dbutils.fs.mkdirs(    "/Volumes/workspace/default/data_volume/archive")


**3. Copy file:**

dbutils.fs.cp(    "/Volumes/workspace/default/data_volume/orders.csv",    "/Volumes/workspace/default/data_volume/archive/orders.csv")


**4. Move file:**

dbutils.fs.mv(    "/Volumes/workspace/default/data_volume/orders.csv",    "/Volumes/workspace/default/data_volume/archive/orders.csv")


**5. Delete file:**

dbutils.fs.rm(    "/Volumes/workspace/default/data_volume/archive/orders.csv")



**Real-time example:**

Before processing incoming CSV files, I use dbutils.fs.ls() to check whether the expected files are available in the landing directory.


**Interview tip:** 

- Use supported Unity Catalog Volume paths for governed file access. 
- Don't assume legacy DBFS root or FileStore is enabled.


## Q15. What is dbutils.notebook.run()?


**dbutils.notebook.run()** is used to `execute another notebook programmatically` and `pass parameters to it`.



**Syntax:**

dbutils.notebook.run(    path,    timeout_seconds,    arguments)


**Parent notebook:**


```python
result = dbutils.notebook.run(
    "./silver_transformation",
    600,
    {
        "source_table": "workspace.ecommerce_bronze.orders",
        "target_table": "workspace.ecommerce_silver.orders"
    }
)

print(result)


**Child notebook:**


```python
# Create notebook widgets
dbutils.widgets.text("source_table", "")
dbutils.widgets.text("target_table", "")

# Get parameter values
source_table = dbutils.widgets.get("source_table")
target_table = dbutils.widgets.get("target_table")

# Read source Bronze table
df = spark.table(source_table)

# Write transformed data into Silver table
(
    df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_table)
)

# Return execution status to parent notebook
dbutils.notebook.exit("Transformation completed")


**Real-time example:**

I use notebook execution to pass Bronze and Silver table names dynamically to a reusable transformation notebook.


**Interview follow-up:**

**What happens if the child notebook fails?**

The parent notebook receives an exception. We can handle it using try-except, log the error, and propagate the failure when the overall pipeline must stop.


## SECTION 4: DATABRICKS WIDGETS


## Q16. What are Databricks Widgets?


- **Databricks Widgets** are `interactive input controls` used to `pass parameters into notebooks`.
- They help create` reusable notebooks without hardcoding values`.


| Widget | Purpose |
|---|---|
| Text | Accept text input |
| Dropdown | Select one predefined value |
| Combobox | Select or enter a value |
| Multiselect | Select multiple values |


**Real-time example:**

In my E-Commerce project, I use widgets to select the source table and environment instead of creating separate notebooks for every table.


## Q17. How do you create and use Widgets?


**1. Text widget:**


```python
dbutils.widgets.text(
    "catalog_name",
    "workspace",
    "Catalog"
)

catalog = dbutils.widgets.get("catalog_name")

print(catalog)



**2. Dropdown widget:**


```python
dbutils.widgets.dropdown(
    "environment",
    "dev",
    ["dev", "test", "prod"]
)

environment = dbutils.widgets.get("environment")

print(environment)


**3. Combobox widget:**


```python
dbutils.widgets.combobox("layer","bronze",["bronze", "silver", "gold"])
layer = dbutils.widgets.get("layer")


**4. Multiselect widget:**


```python
dbutils.widgets.multiselect("tables","customers", ["customers", "orders", "products", "payments"])

tables = dbutils.widgets.get("tables")print(tables)


**Real-time example:**

I use a dropdown widget to select dev, test, or prod, and then use that parameter to select the corresponding pipeline configuration.


## Q18. Explain a real-time parameterized notebook.


**Parameterization** means `passing multiple values` into a `notebook` instead of hardcoding them.


- Suppose we have seven E-Commerce datasets. 
- Rather than creating seven separate ingestion notebooks, we create one reusable notebook and pass the table name dynamically.


### Step 1: Create widgets


```python
dbutils.widgets.text("table_name","orders")
dbutils.widgets.dropdown("environment","dev",
["dev", "test", "prod"])




### Step 2: Retrieve parameters




```python
table_name = dbutils.widgets.get("table_name")
environment = dbutils.widgets.get("environment")


### Step 3: Configure source path


```python
BASE_PATH = ("/Volumes/workspace/default/data_volume/ecommerce_lakehouse")
source_path = f"{BASE_PATH}/landing/{table_name}.csv"


### Step 4: Read CSV


```python
df = (spark.read.option("header", True)   
.option("inferSchema", True)   
.csv(source_path))


### Step 5: Write Bronze table


```python
target_table = f"workspace.ecommerce_bronze.{table_name}"
(df.write.format("delta")
.mode("append")
.saveAsTable(target_table))


**Real-time example:**

The same notebook can process customers, orders, products, and payments by changing the widget parameter.


**Important production consideration:**

- The example demonstrates parameterization, not a complete production ingestion framework. 
- Validate the table name against an approved list and implement duplicate prevention and incremental processing before using append mode in production.


## SECTION 5: DATABRICKS SECRETS


## Q19. What are Databricks Secrets?


- **Databricks Secrets** provide a **secure mechanism** for `storing` and `retrieving sensitive information` such as `passwords, API keys, access tokens,` and `database credentials`.
- Instead of hardcoding credentials inside notebooks, we store them in `secret scopes` and `retrieve them` during execution.


**Important terminology:**


| Term | Meaning |
|---|---|
| Secret Scope | Logical container for secrets |
| Secret | Sensitive key-value information |
| Secret Key | Name used to retrieve a secret |
| ACL | Access Control List for managing permissions |



**Real-time example:**

In an E-Commerce data pipeline, if I need to connect to an external PostgreSQL database, I store the database username and password in Databricks Secrets.


## Q20. How do you create a Secret Scope?


We can create secret scopes using the **Databricks CLI**.

### Step 1: Configure CLI authentication

Install and authenticate the Databricks CLI using an approved authentication method.

### Step 2: Create scope

databricks secrets create-scope ecommerce-scope

### Step 3: Add database username

databricks secrets put-secret ecommerce-scope db-username

### Step 4: Add database password

- databricks secrets put-secret ecommerce-scope db-password
- The CLI supports entering the secret values through its interactive input flow.

### Step 5: Grant permissions

databricks secrets put-acl ecommerce-scope data-engineering-group READ


**Real-time example:**

- I create an ecommerce-scope and store PostgreSQL credentials inside it. 
- Authorized pipeline users can retrieve those credentials without storing them directly in notebook source code.


**Interview tip:**

Secret scope permissions should follow least-privilege principles.


## Q21. How do you retrieve Secrets in a Databricks Notebook?


We use `dbutils.secrets.get()` to **retrieve stored secret values**.



**Example:**


```python
db_username = dbutils.secrets.get(scope="ecommerce-scope",   key="db-username")db_password = dbutils.secrets.get(   scope="ecommerce-scope",    key="db-password")


### How do you verify the secret?

- print("Username retrieved successfully")print("Password retrieved successfully")
- Do not print the actual secret value.


**Real-time example:**

Before executing a JDBC read operation, I retrieve the database credentials from the secret scope and pass them into the JDBC connection properties.


**Interview tip:**

Secret values may be redacted in supported outputs, but redaction does not prevent an authorized user from retrieving them.


## Q22. How do you use Databricks Secrets in a JDBC connection?


- **JDBC** stands for `Java Database Connectivity`. 
- It allows `Spark` to connect to` external relational databases` such as `PostgreSQL, MySQL,` and `SQL Server` through supported JDBC drivers.
- Instead of `hardcoding credentials`, I retrieve them from `Databricks Secrets`.


### Step 1: Retrieve credentials


```python
db_username = dbutils.secrets.get(scope="ecommerce-scope",   key="db-username")db_password = dbutils.secrets.get(   scope="ecommerce-scope",    key="db-password")


### Step 2: Define JDBC URL


```python
jdbc_url = ("jdbc:postgresql://database-host:5432/ecommerce")


### Step 3: Define connection properties


```python
connection_properties = {"user": db_username,"password": db_password,"driver": "org.postgresql.Driver"}


### Step 4: Read database table


```python
df = spark.read.jdbc(url=jdbc_url,table="customers", properties=connection_properties)display(df)


**Real-time example:**

- Suppose the E-Commerce customer master data is stored in PostgreSQL. 
- I connect to the database using JDBC, retrieve the credentials from Databricks Secrets, and load the customer records into the Bronze layer.


**Important:**

Ensure the PostgreSQL JDBC driver is available and the compute environment has network connectivity to the database.


## SECTION 6: DATABRICKS VOLUMES


## Q23. What are Databricks Volumes?


- **Databricks Volumes** are `Unity Catalog-governed objects` used to `store and manage non-tabular files` such as `CSV, JSON, Parquet, images, and other files`.
- They provide `centralized access control` for `file-based data`.



**Volume hierarchy:**


```text
Catalog
   |
   |-- workspace
          |
          |-- Schema
                 |
                 |-- default
                        |
                        |-- Volume
                               |
                               |-- data_volume
                                      |
                                      |-- customers.csv
                                      |-- orders.csv
                                      |-- products.csv



**Volume path:**

/Volumes/workspace/default/data_volume/



**Real-time example:**

In my E-Commerce Lakehouse project, I use a Unity Catalog Volume as the landing location for seven source CSV datasets before processing them into Bronze Delta tables.


## Q24. What are the types of Databricks Volumes?


Databricks supports **two types** of Unity Catalog Volumes:


**1. Managed Volume:**

- Databricks manages storage location.
- Unity Catalog manages permissions.
- Suitable for files whose storage location is managed through Databricks.

**2. External Volume:**

- References an existing external cloud storage location.
- Unity Catalog governs access to the files.
- Suitable when organizations already maintain cloud storage locations.


| Managed Volume | External Volume |
|---|---|
| Managed storage location | Existing external storage location |
| Databricks manages storage placement | References external storage |
| Unity Catalog permissions | Unity Catalog permissions |


**Real-time example:**

- If my project stores incoming CSV files in a Databricks-managed storage location, I can use a Managed Volume. 
- If the organization already has files in an existing cloud storage location, an External Volume may be appropriate.


## Q25. How do you create a Volume?


We can create a **Volume using SQL**.

### Step 1: Create Volume

CREATE VOLUME IF NOT EXISTS

workspace.default.data_volume;


### Step 2: List files

display(dbutils.fs.ls("/Volumes/workspace/default/data_volume/"))


### Step 3: Read CSV

df = (spark.read.option("header", True)    
.option("inferSchema", True)    
.csv("/Volumes/workspace/default/data_volume/customers.csv"))
display(df)


### Step 4: Check records

print("Total records:", df.count())df.printSchema()


**Real-time example:**

I create a Volume called data_volume inside workspace.default and use it to store the source CSV datasets required by the E-Commerce Lakehouse pipeline.


## Q26. What is the difference between DBFS and Unity Catalog Volumes?



- **DBFS** stands for `Databricks File System`. 
- It is a `legacy filesystem abstraction` used in **older Databricks environments**.
- **Unity Catalog Volumes** are `governed file-storage objects` integrated with `Unity Catalog`.


| DBFS | Unity Catalog Volumes |
|---|---|
| Legacy filesystem abstraction | Unity Catalog-governed file storage |
| Legacy DBFS root may be disabled | Recommended for governed file access |
| Legacy paths and access patterns | Supports Unity Catalog permissions |
| Often found in older projects | Suitable for modern governed architectures |


**Real-time example:**

- In my Databricks environment, DBFS root and FileStore access are restricted. 
- Therefore, I use /Volumes/workspace/default/data_volume/ to store and access project CSV files.


**Interview tip:**

- Do not say DBFS and Volumes are exactly the same. 
- Volumes provide a Unity Catalog governance model for file access.


## SECTION 7: DATABRICKS GIT FOLDERS (REPOS)


## Q27. What are Databricks Repos?


- **Databricks Repos**, now commonly called `Git folders`.
- It provides `Git integration` inside the `Databricks workspace`.
- They allow developers to manage `notebooks` and `project files` using **version control systems** such as `GitHub, GitLab, and Azure DevOps.`


| Operation | Purpose |
|---|---|
| Clone | Create a workspace copy of a remote repository |
| Commit | Save changes in local Git history |
| Push | Upload committed changes to remote repository |
| Pull | Download and integrate remote changes |
| Branch | Create an independent development line |
| Merge | Combine changes from different branches |



**Real-time example:**

I connect my Data Engineering Python GitHub repository to Databricks Git folders so I can manage notebook changes and maintain version history.


## Q28. How do you connect GitHub to Databricks?


**Follow these steps:**

1. Open Databricks Workspace.
2. Navigate to Git folders.
3. Click Create Git folder.
4. Select GitHub as the Git provider.
5. Authenticate using configured Git credentials.
6. Enter the GitHub repository URL.
7. Select the required branch.
8. Click Create.
Example repository:
https://github.com/franklinsurya4/Data_Engineering_Python


**Real-time example:**

I connect my GitHub repository to Databricks so I can maintain notebooks in a version-controlled environment instead of keeping all project code only inside the workspace.


## Q29. Explain the Git folder workflow.


The **Git folder workflow** allows `developers to maintain and synchronize project code` between **Databricks** and **GitHub**.


**Workflow:**


```text
GitHub Repository
       |
       v
Clone Repository
       |
       v
Databricks Git Folder
       |
       v
Create / Modify Notebook
       |
       v
Review Changes
       |
       v
Commit Changes
       |
       v
Push Changes
       |
       v
GitHub Updated


### Step 1: Clone

Connect the `remote GitHub repository` to Databricks.

### Step 2: Develop

`Create or modify notebooks` inside the Git folder.

### Step 3: Commit

`Save the changes` with a **meaningful commit message**.

**Example:**

Added reusable Bronze ingestion notebook

### Step 4: Push

`Upload committed changes` to GitHub.

### Step 5: Pull

`Download changes` made by other developers.

### Step 6: Merge

Merge approved changes into the main branch using the team's Git workflow.


**Real-time example:**

I develop a Silver transformation notebook, commit the changes, and push them to GitHub so that the updated notebook is available in the remote repository.


## Q30. What is the difference between Workspace notebooks and Git folder notebooks?


- **Workspace notebooks** are maintained as `Databricks workspace artifacts`. 
- **Git folder notebooks** are maintained inside `Git-integrated folders`.




| Workspace Notebooks | Git Folder Notebooks |
|---|---|
| Stored in Databricks Workspace | Stored in Git-backed folders |
| Not automatically version-controlled by Git | Git version control supported |
| Useful for workspace development | Useful for collaborative development |
| Changes are not automatically pushed to GitHub | Changes can be committed and pushed |
| Workspace permissions apply | Workspace and Git permissions apply |


**Real-time example:**

I use regular workspace notebooks for experimentation and Git folders for maintaining reusable project notebooks with version control.


**Interview tip:**

Editing a notebook inside a Git folder does not automatically push it to GitHub. You must commit and push your changes.


## SECTION 8: DAY-TO-DAY DATABRICKS USAGE


## Q31. Explain your daily activities as a Databricks Data Engineer.


In my day-to-day work, I use Databricks to develop, execute, monitor, and maintain data engineering pipelines.


**My typical activities include:**

- Opening the Databricks Workspace and accessing project notebooks.
- Connecting to the required compute environment.
- Reading incoming CSV or Parquet files from Unity Catalog Volumes.
- Using PySpark to perform data cleaning and transformations.
- Writing processed data into Bronze, Silver, and Gold Delta tables.
- Using widgets to parameterize notebooks.
- Retrieving credentials using Databricks Secrets.
- Using Git folders to commit and push notebook changes.
- Configuring Databricks Workflows for automated execution.
- Monitoring job execution and troubleshooting failures.
- Validating output tables using Spark SQL.
- Supporting dashboard requirements through Gold-layer datasets.


**Real-time example:**

- In my E-Commerce Lakehouse project, I ingest incoming customer, product, order, payment, and shipment datasets. 
- I process the data through Bronze, Silver, and Gold layers, validate the output, and use Databricks SQL to generate business reports.


## Q32. Explain how you set up and parameterize a notebook.


**Notebook parameterization** allows us to `pass multiple values` into a `notebook` without `changing its source code`.


- Suppose we have seven E-Commerce source datasets. 
- Instead of creating seven ingestion notebooks, I create one reusable notebook.


### Step 1: Create widgets


```python
dbutils.widgets.text("source_file","orders.csv")
dbutils.widgets.text("target_table","orders")
dbutils.widgets.dropdown("environment","dev",   
["dev", "test", "prod"])


### Step 2: Retrieve parameters


```python
source_file = dbutils.widgets.get("source_file")
target_table = dbutils.widgets.get("target_table")
environment = dbutils.widgets.get("environment")


### Step 3: Configure source path


```python
BASE_PATH = ("/Volumes/workspace/default/data_volume/ecommerce_lakehouse")
source_path = f"{BASE_PATH}/landing/{source_file}"


### Step 4: Read source data


```python
df = (spark.read.option("header", True)    
                .option("inferSchema", True)    
                .csv(source_path))


**Real-time example:**

- The same notebook can process customers.csv, orders.csv, and products.
- csv by passing different widget values.


**Production consideration:**

- Add schema validation, ingestion metadata, and idempotent processing. 
- Do not repeatedly append the same files without duplicate prevention.


## Q33. How do you connect a Git folder and push a notebook?


First, I connect my GitHub repository to Databricks using Git folders.


**Then I follow these steps:**

- Create or open a Git folder.
- Select the required branch.
- Create or modify the notebook.
- Review the changed files.
- Enter a meaningful commit message.
- Commit the changes.
- Push the committed changes to GitHub.
- Verify the updated notebook in the remote repository.


**Example commit message:**

Added parameterized Bronze ingestion notebook.


**Real-time example:**

After developing the E-Commerce Bronze ingestion notebook, I commit and push it to GitHub so that the project source code is version-controlled.


**Interview follow-up: **

**What happens if two developers modify the same notebook?**

A Git merge conflict may occur if their changes overlap. I compare the conflicting changes, resolve them, test the updated notebook, and commit the resolved version.


## SECTION 9: SCENARIO-BASED INTERVIEW QUESTIONS


## Q34. Your notebook works on All-purpose compute but fails on Serverless compute. What will you check?


I first check whether the notebook uses any `APIs, libraries, configurations, or filesystem operations` that are `unsupported in the serverless environment`.


**I investigate:**

- Unsupported Spark APIs.
- RDD-related operations.
- Restricted filesystem access.
- Unity Catalog permissions.
- Python package compatibility.
- Network connectivity.
- Unsupported Spark configurations.


**Real-time example:**

- Suppose my notebook uses an RDD-based transformation. It works on a compatible classic cluster but fails on a serverless environment that does not support that RDD access pattern.
- I would rewrite the transformation using DataFrame APIs wherever possible and test it again.


**Interview tip:**

Never assume every classic-cluster feature works identically in serverless compute.


## Q35. How do you prevent passwords from appearing in notebooks?


I use Databricks Secrets to securely store sensitive credentials instead of hardcoding passwords inside notebooks.


**My approach:**

1. Create a secret scope.
2. Store credentials as secrets.
3. Configure appropriate ACL permissions.
4. Retrieve credentials using dbutils.secrets.get().
5. Pass credentials to the JDBC connection.
6. Avoid printing credentials in notebook outputs or logs.


**Code:**


```python
password = dbutils.secrets.get(scope="ecommerce-scope",  key="db-password")


**Real-time example:**

When connecting to an external PostgreSQL database, I retrieve the password from the secret scope instead of writing it directly in the notebook.


Instead of maintaining separate notebooks for each table, I create a reusable notebook.


## Q36. Your notebook needs to process different tables every day. How do you handle it?


I use notebook parameterization with Databricks Widgets and pass the required table name through Databricks Workflows.


**Example:**


```python
dbutils.widgets.text("table_name", "orders")
table_name = dbutils.widgets.get("table_name")
source_table = f"workspace.ecommerce_bronze.{table_name}"
df = spark.table(source_table)display(df)


**Real-time example:**

- The same transformation notebook can process customer, order, payment, and shipment tables by changing the parameter.
- For production, I validate the table name against an approved list before using it in SQL identifiers.


## Q37. How do you share notebooks between developers?


I use Databricks Git folders integrated with GitHub.


**My workflow is:**

- Connect the repository.
- Create a feature branch.
- Develop notebooks.
- Commit the changes.
- Push changes to the remote repository.
- Create a pull request.
- Review and merge approved changes.


**Real-time example:**

If one developer works on Bronze ingestion and another works on Silver transformation, both can develop their changes in separate branches and integrate them through pull requests.


**Interview tip:**

- Git provides version control; 
- Unity Catalog and workspace permissions govern data and workspace access. 
- They solve different problems.


## Q38. How do you manage CSV files in Databricks?


I use Unity Catalog Volumes to store `incoming CSV files` and `Spark` or `Auto Loader` to process them.


**My typical workflow:**

1. Upload or receive CSV files in the landing location.
2. Read the files using Spark or Auto Loader.
3. Validate the schema.
4. Check missing and invalid records.
5. Add ingestion metadata.
6. Write the records into Bronze Delta tables.
7. Apply transformations in Silver.


**Example:**


```python
df = (spark.read.option("header", True)    
.option("inferSchema", True)    
.csv("/Volumes/workspace/default/data_volume/orders.csv"))

display(df)



**Real-time example:**

I receive orders.csv in the E-Commerce landing folder and ingest it into the Bronze orders table before performing data-quality checks.


## Q39. A developer accidentally deletes a file from a Volume. How would you recover it?


- First, I check whether the original source file is available in the upstream system, backup location, or archive.
- If the file is unavailable, I check whether cloud-storage versioning, backup, or recovery mechanisms were configured.
- If a recoverable version exists, I restore it to the Volume and verify its contents.


**Real-time example:**

Suppose orders.csv is accidentally deleted from the landing Volume. I check the original source delivery location or configured backup. Once recovered, I validate the file and rerun the ingestion process.


**Important interview point:** 

Unity Catalog permissions provide governance, but they do not automatically guarantee file recovery.


## Q40. What happens when a Databricks cluster terminates?


When a Databricks cluster terminates, its compute resources are released and Spark processing stops.

**The following things happen:**

1. Running Spark jobs stop or fail.
2. In-memory cached DataFrames are lost.
3. Driver and executor processes terminate.
4. Compute resources are released.
5. Workspace notebooks remain available.
6. Persistent Delta tables remain stored in cloud storage.
7. Data stored in Unity Catalog Volumes remains available.


**Real-time example:**

- Suppose I process 10 million E-Commerce order records and save the output into a Silver Delta table. 
- If the cluster terminates after successful execution, the Silver Delta table remains available because it is stored persistently, not in cluster memory.



**Interview tip:** 

Cluster termination does not mean persistent data deletion.
