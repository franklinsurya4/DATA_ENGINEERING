
## Q1. What is a SQL Warehouse?


A **SQL Warehouse** is a `Databricks compute resource` designed to execute `SQL queries for data analytics, dashboards, BI tools` and `reporting`.


## Q2. What is the difference between a SQL Warehouse and Serverless Spark compute?


- A **SQL Warehouse** is a `Databricks compute resource` designed to execute `SQL queries for data analytics, dashboards, BI tools` and `reporting`.
-  A **Serverless Spark compute** is used for `broader data engineering workloads` such as` PySpark transformations, ETL` and `data processing`.


**In my E-Commerce project:**

-   I use Spark compute to clean Bronze data and build Silver and Gold tables. 
-   I use a SQL Warehouse to query Gold tables and create dashboards.



## Q3. What is MPP?


- **MPP** stands for `Massively Parallel Processing.` 
- It is an architecture in which a `large query` is divided into `smaller tasks` and 
- processed `concurrently` across `multiple compute resources`.

**For example:**

-   a query on 100 million sales records can be divided into smaller tasks, processed in parallel and combined to produce the final result.


## Q4. What is Photon?


- A **Photon** is `Databricks' native vectorized query execution engine`. 
- It accelerates supported` SQL and DataFrame operations` using `native C++ execution` and `columnar batch processing`.
- It improves the performance of s**upported scans, joins** and **aggregations**.


## Q5. Is Photon the same as Apache Spark?


- No. 
- **Apache Spark** is a `distributed processing framework`,
- **Photon** is `Databricks' optimized execution engine` that   accelerates supported **Spark SQL** and **DataFrame operations**.


## Q6. What is a Dimension table?


A **Dimension table** stores `descriptive business information` used to `filter, group` and `analyze data`.

**For example:**

-    a customer dimension can contain customer ID, city, state and customer segment.


## Q7. What is a Fact table?


A **Fact table** stores `measurable business events` or `transactions`, usually along with` foreign keys that connect to dimension tables`.

**For example:**

-   an order sales fact table can contain order ID, customer ID, product ID, quantity and sales amount.


### Q8. What is a Measure?

A Measure is a numeric value that can be aggregated using calculations such as SUM, COUNT, AVG, MIN and MAX.

Examples include sales amount, quantity and order count.


### Q9. What is a Metric?

A Metric is a defined measurement used to evaluate business or operational performance.

For example, Average Order Value (AOV) measures the average revenue generated per order.


## Q10. What is a KPI?


- KPI stands for `Key Performance Indicator`. 
- It is a key metric used to `measure progress toward a specific business objective`.

**For example:** 

-   achieving ₹10 lakh in monthly revenue can be a business KPI.


## Q11. What is Reporting?


**Reporting** is the process of `presenting business information` in a `structured format` to help `users understand performance` and `make decisions`.

**For example:**

-    a monthly sales report showing revenue, orders and product performance.


## Q12. What is Data Visualization?


**Data Visualization** is the `graphical representation of data` using `charts, graphs, maps` and `other visual elements`.

**For example:**

- a bar chart showing revenue by product category.


## Q13. What is a Dashboard?


A **Dashboard** is a `collection of visualizations, reports, KPIs` and `filters` that provides a `consolidated view of business performance`.

**For example:**

- an E-Commerce Executive Dashboard showing revenue, orders, payment performance and inventory status.


### Q14. What is Drill Down?


**Drill Down** means `moving from summarized information to more detailed information`.

**For example:**

Year → Quarter → Month → Day.


## Q15. What is Drill Up?


**Drill Up** means `moving from detailed information to summarized information`.

**For example:**

Day → Month → Quarter → Year.

## Q25. What is a SQL Alert in Databricks?


A **SQL Alert** is used to monitors the `result of a SQL query` and `sends a notification` when a configured `condition is met`.

**For example:**

- I can create an alert when the payment failure percentage exceeds 5%.


## Q26. How do you monitor data quality using SQL Alerts?


- I create SQL queries to identify data quality issues such as null values, duplicate records and invalid records. 
- Then I configure alerts with appropriate conditions and thresholds so the responsible team is notified when an issue occurs.


## Q27. How do you create an alert for invalid orders?


**I execute a SQL query that counts records with null order IDs:**


```sql
SELECT
    COUNT(*) AS invalid_order_count
FROM workspace.ecommerce_silver.orders
WHERE order_id IS NULL;
```


Then I configure a SQL Alert to trigger when `invalid_order_count` is greater than zero.


## Q28. How do you monitor payment failure rates using SQL Alerts?


I calculate the percentage of failed payments using a SQL query and configure an alert when the result exceeds an agreed business threshold.


```sql
SELECT
    ROUND(
        100.0 * SUM(
            CASE
                WHEN payment_status = 'FAILED' THEN 1
                ELSE 0
            END
        ) / NULLIF(COUNT(*), 0),
        2
    ) AS payment_failure_percentage
FROM workspace.ecommerce_gold.payment_metrics;
```


- This example assumes the table contains individual payment records. 
- If it contains aggregated metrics, I use the actual payment count columns instead.


## Q29. How do you monitor low inventory using SQL Alerts?


I write a SQL query to count products whose available quantity is below the defined stock threshold.


```sql
SELECT
    COUNT(*) AS low_stock_products
FROM workspace.ecommerce_gold.inventory_status
WHERE available_quantity < 10;
```


Then I configure an alert when the result is greater than zero.


## Q30. Can a Databricks SQL Alert monitor job failures?


- Yes. 
- Job failures can be monitored using Lakeflow Jobs failure notifications or SQL Alerts based on supported job-monitoring data.
- A SQL Alert does not automatically monitor every job; it needs a suitable SQL query and condition.


## Q31. What is Genie in Databricks?


Genie Agents allow business users to ask questions about governed data using natural language instead of writing SQL queries.

**For example:**

-   a user can ask, "What is the total revenue this month?" 
    Genie can generate SQL, execute the query and return results and visualizations.



## Q32. What is the difference between Genie and a Dashboard?


- A **Dashboard** displays` predefined visualizations, KPIs` and `reports` 
- **Genie** allows users to `explore data` by **asking questions in natural language**.

## Q35. What are business instructions in Genie?


**Business instructions** defines `how Genie should understand business terminology`, `use datasets`, `construct SQL queries` and `calculate metrics`.

**For example:**

-    I can define Average Order Value as total revenue divided by total orders and instruct Genie to use the existing Gold table definitions.


## Q36. What are trusted example SQL queries in Genie?


- **Trusted example SQL queries** are` verified SQL examples associated` with business questions. 
- They help Genie understand the `expected query patterns` and improve the `accuracy of generated SQL`.

**For example:**

-    I can provide a tested query for identifying the top 10 products by revenue.


## Q37. Does Genie always generate correct SQL?


- No. 
- Genie can generate incorrect SQL if table relationships, column definitions or business instructions are unclear.
- Therefore, I validate the generated SQL, check the table schemas and verify the results before using them for business decisions.


## Q38. How do you protect sensitive data in Databricks dashboards?


- I use `Unity Catalog permissions, row filters, column masks` and `secured views where appropriate`. 
- I also restrict **dashboard datasets** to the `columns` and `records` required for reporting.
- Dashboard filters alone do not provide data security.


## Q39. What is the difference between a Dashboard, SQL Alert and Genie?


- A **Dashboard** is used for `visual reporting`, 
- A **SQL Alert** is used for `monitoring query results against defined conditions`, 
- A **Genie** is used for `conversational analytics` through natural-language questions.
