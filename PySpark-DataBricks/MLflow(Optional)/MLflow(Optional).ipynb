
## Q1. What is MLflow?



MLflow is an open-source platform used to manage the machine learning lifecycle.


**It helps teams:**

1. Track machine learning experiments.
2. Record parameters and metrics.
3. Store model artifacts.
4. Register and version models.
5. Deploy models for predictions.


**Real-time example:** 

A retail company uses MLflow to compare different customer churn prediction models and manage the selected model.


## Q2. What is an MLflow Experiment?


An experiment is a logical container used to organize related machine learning runs.


**Example:** 

A customer churn prediction experiment may contain multiple runs using different algorithms and configurations.


```python
import mlflow

mlflow.set_experiment("/Shared/customer_churn_experiment")
```


## Q3. What is an MLflow Run?


An MLflow run represents one execution of model-training code.


**Each run can record:**

- Parameters
- Metrics
- Artifacts
- Model information


**Example:** 

Training a Random Forest model with `n_estimators=100` creates a run containing its configuration and evaluation results.


## Q4. What is Experiment Tracking?


Experiment tracking records and compares different model-training executions.


**Real-time example:** 

A data scientist trains three customer churn models and compares their accuracy to identify the best-performing model.


```python
with mlflow.start_run(run_name="random_forest_v1"):

    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("max_depth", 5)

    mlflow.log_metric("accuracy", 0.92)
```


## Q5. What are Parameters and Metrics in MLflow?




| Parameters | Metrics |
|---|---|
| Inputs or configurations used during training | Measurements used to evaluate model performance |
| Usually selected before training | Calculated during or after training |
| `max_depth=5` | `accuracy=0.92` |
| `learning_rate=0.01` | `precision=0.89` |


**Example:**


```python
mlflow.log_param("max_depth", 5)
mlflow.log_metric("accuracy", 0.92)
```


## Q6. What are Artifacts in MLflow?


Artifacts are files generated during a machine learning experiment.



**Examples:**

1. Trained model files.
2. Confusion matrix images.
3. Evaluation reports.
4. Feature importance plots.
5. CSV files.


```python
mlflow.log_artifact("evaluation_report.csv")
```


## Q7. What is the difference between an Experiment and a Run?


| Experiment | Run |
|---|---|
| Organizes related training activities | Represents one training execution |
| Contains multiple runs | Belongs to an experiment |
| Example: Customer churn experiment | Example: Random Forest with depth 5 |

## Q9. What is MLflow Model Registry?


Model Registry is a centralized system for registering, versioning and managing machine learning models.


**Real-time example:**

- **Version 1**: Initial fraud detection model.
- **Version 2**: Improved model with additional transaction features.
- **Version 3**: Updated model with better evaluation results.


In Databricks, Unity Catalog is the recommended approach for model governance and registry management.


**Example model name:**


```text
workspace.ecommerce_gold.customer_churn_model
```


**This follows the three-level naming convention:**


```text
catalog.schema.model_name
```


## Q10. What is Unity Catalog Model Registry?


Unity Catalog Model Registry manages registered models using centralized governance.


**It provides:**

1. Access control.
2. Model versioning.
3. Model lineage.
4. Model discovery.
5. Auditing.
6. Cross-workspace model access.


**Real-time example:** 

A machine learning team registers a fraud detection model in Unity Catalog and grants the production team permission to use it.


## Q11. What is Model Serving in Databricks?


Model Serving deploys a trained model behind an endpoint so applications can send input data and receive predictions through a REST API.


**Real-time example:**

- An e-commerce application sends transaction details to a fraud detection endpoint. 
- The deployed model returns whether the transaction is potentially fraudulent.



```text
Application
    |
    v
REST API Endpoint
    |
    v
Deployed ML Model
    |
    v
Prediction
```



## Q12. What is the difference between Model Registry and Model Serving?


| Model Registry | Model Serving |
|---|---|
| Manages registered models | Deploys models for predictions |
| Maintains model versions | Exposes models through endpoints |
| Supports governance and lineage | Supports real-time inference |
| Example: Fraud model Version 2 | Example: Fraud prediction API |


## Q14. How does MLflow help with model reproducibility?


MLflow records information about model-training runs, including parameters, metrics, artifacts and model information.


**This helps teams:**

- Compare previous experiments.
- Identify which configuration produced a model.
- Reproduce training workflows.
- Trace a registered model back to its source run.


**Real-time example:** 

If a fraud detection model's accuracy decreases, the team can inspect its earlier runs and compare the training configurations.


## Q15. Explain the MLflow workflow in Databricks.


**A typical MLflow workflow includes:**

1. Prepare and process the training dataset.
2. Train a machine learning model.
3. Track experiments using MLflow.
4. Log parameters, metrics and artifacts.
5. Compare different runs.
6. Register the selected model in Unity Catalog.
7. Deploy the registered model using Model Serving.
8. Monitor prediction performance.



**E-Commerce example:** 

Train a customer churn model using customer purchase history, compare experiment runs, register the selected model and expose it through a serving endpoint.


## Q16. How is MLflow useful for a Data Engineer?


MLflow helps Data Engineers collaborate with Data Scientists and ML Engineers by supporting model-training pipelines and tracking.


**A Data Engineer may:**

1. Build data pipelines to prepare training datasets.
2. Validate and transform source data.
3. Provide clean Gold-layer tables for machine learning.
4. Schedule data preparation and model-training workflows.
5. Support model monitoring and retraining pipelines.



**E-Commerce example:** 

- A Data Engineer prepares customer purchase history in a Gold table.
- A Data Scientist uses that dataset to train a customer churn prediction model and tracks the experiments using MLflow.
