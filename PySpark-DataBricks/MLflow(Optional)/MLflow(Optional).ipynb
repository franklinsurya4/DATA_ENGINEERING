MLflow is an open-source platform used to manage the machine learning lifecycle. In Databricks, it helps data scientists and ML engineers track experiments, compare model performance, manage model versions, and deploy models.

For your Data Engineering interview, understand the concepts and perform one basic experiment. You do not need to master machine learning algorithms for this topic.
15.1 Important MLflow concepts
Concept	Meaning	Real-time example
Experiment Tracking	Records model training experiments	Compare different models for predicting customer churn
Parameters	Inputs or settings used during training	max_depth=5, n_estimators=100
Metrics	Numerical measurements of model performance	Accuracy = 0.92
Artifacts	Files generated during model training	Model file, plots, CSV reports
Model Registry	Manages registered model versions and lifecycle	Maintain versions 1, 2 and 3 of a fraud detection model
Model Serving	Makes a trained model available for predictions through an endpoint	Predict whether a new transaction is fraudulent

15.6 Interview questions and answers
Q1. What is MLflow?

MLflow is an open-source platform for managing the machine learning lifecycle, including experiment tracking, model management and deployment.

Q2. What is an MLflow run?

An MLflow run represents one execution of model-training code. It records parameters, metrics and artifacts.

Q3. What is experiment tracking?

It records and compares different model-training experiments to identify which configuration performs better.

Q4. What is a model registry?

A centralized repository for managing model versions, permissions and lifecycle.

Q5. What is model serving?

It deploys a trained model as an API endpoint to generate predictions from new input data.

Q6. What is the difference between parameters and metrics?
- Parameters: Settings used to train a model, such as max_depth=5.
- Metrics: Measurements of model performance, such as accuracy or precision.
Q7. Why is MLflow useful in Databricks?

It provides integrated experiment tracking and model management, helping teams monitor, compare, govern and deploy models.
