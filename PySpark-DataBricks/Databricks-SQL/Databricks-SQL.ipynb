
1. SQL Warehouse vs Serverless Spark
A SQL Warehouse is a compute resource designed for SQL analytics, dashboards, BI tools and reporting. It is separate from the Spark compute used by notebooks and Spark workloads.
Feature	Serverless SQL Warehouse	Serverless Spark compute
Main purpose	SQL analytics and BI	Data engineering and general Spark processing
Interface	Databricks SQL	Notebooks and supported Spark workloads
Languages	SQL	Python, SQL, Scala where supported
Engine	Photon-enabled SQL execution	Apache Spark with supported Photon acceleration
Typical workload	Dashboards, reports, ad hoc queries	PySpark transformations, ETL, notebooks
Startup	Low startup latency	Managed serverless startup
Best example	Query Gold tables	Clean Bronze data and build Silver tables

Interview answer: I use serverless Spark compute for PySpark-based data engineering and serverless SQL Warehouses for querying Gold tables, dashboards and business reporting.
What is MPP?
MPP stands for Massively Parallel Processing.

It means a large query can be divided into smaller tasks and processed across multiple compute resources in parallel.

Example: Suppose your Gold sales table contains 100 million records. Instead of processing all records sequentially on one machine, the distributed query engine can divide the work, process partitions in parallel and combine the results.
What is Photon?










Photon is Databricks' native vectorized query execution engine. It accelerates SQL and supported DataFrame workloads.

It is built using native C++ execution and processes data in columnar batches.


Traditional row-oriented execution processes records individually. Photon processes batches of columnar data, making better use of modern CPU capabilities.

Benefits:
- Faster SQL query execution for supported operations.
- Improved scans, joins and aggregations.
- Better performance for many BI and ETL workloads.
- No need to rewrite supported SQL queries to use Photon.
Photon is enabled by default for SQL Warehouses. Unsupported operations can fall back to the Spark execution engine. Photon is not a replacement for Spark or a separate storage system.

Interview question: Is Photon the same as Spark?

No. Apache Spark is the distributed processing framework. Photon is Databricks' optimized execution engine that accelerates supported Spark SQL and DataFrame operations.

5. Databricks SQL Alerts
An Alert monitors a SQL query result and sends a notification when a defined condition is met.

Example business requirements:
- Alert if payment failures exceed 5%.
- Alert if invalid orders appear in Silver.
- Alert if inventory falls below a minimum stock threshold.
- Alert if daily sales fall below an agreed SLA.
- Alert if a pipeline's operational status indicates failure.
The alert executes a query on a schedule, evaluates its condition and sends notifications when the condition is satisfied.

5.2 Build an alert for data quality
Let's monitor null order_id values in your Silver orders table.

First, create this SQL query:
SELECT
    COUNT(*) AS invalid_order_count
FROM workspace.ecommerce_silver.orders
WHERE order_id IS NULL;


Expected healthy result
The query should return 0 when all orders have valid order IDs.
5.3 Configure the alert

2. Save it as Orders_Data_Quality_Check.
3. Open Alerts in the Databricks sidebar.
4. Create a new alert using the saved query.
5. Select your SQL Warehouse.
6. Set the monitored value to invalid_order_count.
7. Set the condition to greater than 0.
8. Test the alert condition.

5.4 More useful E-Commerce alerts
Alert A — Payment failure rate

Assuming payment_status contains SUCCESS and FAILED:
SELECT
    ROUND(
        100.0 * SUM(
            CASE
                WHEN payment_status = 'FAILED' THEN 1
                ELSE 0
            END
        ) / NULLIF(COUNT(*), 0),
        2
    ) AS payment_failure_percentage
FROM workspace.ecommerce_gold.payment_metrics;


Configure a threshold appropriate to the business, such as greater than 5%.

Check the table's grain before using this calculation: if payment_metrics is already aggregated, you must use its actual count columns rather than counting rows as individual payments.

Alert B — Low inventory

Assuming your inventory table has available_quantity:
SELECT
    COUNT(*) AS low_stock_products
FROM workspace.ecommerce_gold.inventory_status
WHERE available_quantity < 10;


Configure the alert when the result is greater than zero.

Alert C — Missing daily sales

Assuming order_date is a date column:
SELECT
    COUNT(*) AS missing_sales_days
FROM workspace.ecommerce_gold.daily_sales
WHERE order_date = CURRENT_DATE();


Configure an alert when the result is zero, if the Gold daily-sales table is expected to contain a row every day. If no orders can legitimately occur on some days, this check should instead validate ingestion completeness.

Alert D — Job failure monitoring

For job failures, use Lakeflow Jobs failure notifications or query supported system tables and create a SQL alert. A SQL alert does not automatically monitor every job; it must query a suitable status source.

8. Interview questions and answers
Q1. What is Databricks SQL?

Databricks SQL provides SQL-based analytics and data warehousing capabilities on the Databricks Lakehouse. It supports SQL Warehouses, queries, dashboards, alerts and Genie.

Q2. What is a SQL Warehouse?

It is a compute resource used to execute SQL queries against Unity Catalog tables and other supported data sources.

Q3. What is the difference between a SQL Warehouse and a Spark cluster?

A SQL Warehouse is optimized for SQL analytics and BI workloads. Spark compute supports broader data engineering and processing workloads, including PySpark.

Q4. What is Photon?

Photon is Databricks' native vectorized execution engine. It accelerates supported SQL and DataFrame operations using columnar batch processing.

Q5. What is MPP?

Massively Parallel Processing distributes query operations across multiple compute resources so they can execute concurrently.

Q6. What is an AI/BI Dashboard?

It is a Databricks business intelligence feature that combines datasets, visualizations, filters and reports to help users understand business performance.

Q7. What is the difference between a metric and a KPI?

A metric measures something. A KPI is a key metric selected to measure progress toward a specific business objective.

Q8. What is drill down?

Drill down means moving from summarized information to more detailed information, such as year → quarter → month → day.

Q9. What is a Databricks SQL Alert?

An alert periodically evaluates a SQL query result against a configured condition and sends a notification when the condition is met.

Q10. How do you monitor data quality using alerts?

I create SQL queries to measure invalid records, missing values or duplicates. I configure alerts with appropriate thresholds and schedules so the responsible team is notified when data quality deteriorates.

Q11. What is Genie?

Genie is a natural-language analytics capability that allows users to ask questions about governed data and receive generated SQL, results and visualizations.

Q12. How do you protect sensitive data in dashboards?

I apply Unity Catalog permissions, row filters and column masks to the underlying data. I also restrict dashboard datasets to the columns and rows needed.

Q13. How do you build an E-Commerce dashboard?

I use Gold tables as dashboard datasets, select an appropriate SQL Warehouse, create KPIs and visualizations, add filters, validate the calculations and publish the dashboard with appropriate access controls.

Q14. Can an alert monitor a job failure?

Yes, through job failure notifications or a SQL alert that queries supported job monitoring data. A SQL alert needs a query and condition; it does not automatically monitor all jobs.

Q15. Why use Gold tables for dashboards?

Gold tables contain curated business-ready data and aggregated metrics. They reduce repeated transformation work and provide consistent reporting sources.
