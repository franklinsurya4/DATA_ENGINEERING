
SECTION 1: DATABRICKS PLATFORM FUNDAMENTALS
Q1. What is Databricks?
Answer:

Databricks is a cloud-based data intelligence platform built around Apache Spark. It provides a unified environment for data engineering, data analytics, machine learning, and AI workloads.

It supports Lakehouse architecture, which combines the scalability of data lakes with the reliability and management capabilities of data warehouses.

Key features:
1. Apache Spark for distributed data processing.
2. Delta Lake for reliable data storage.
3. Unity Catalog for data governance and access control.
4. Databricks Notebooks for development.
5. Databricks Workflows for pipeline orchestration.
6. Databricks SQL for analytical queries.
7. Auto Loader for incremental file ingestion.
Real-time example:

In my E-Commerce Lakehouse project, I use Databricks to ingest seven CSV datasets, process them using PySpark, and store the results in Bronze, Silver, and Gold Delta tables.
- Bronze: Raw data ingestion.
- Silver: Data cleaning and validation.
- Gold: Business-level aggregations.
- Workflows: Pipeline orchestration.
- Unity Catalog: Data governance.
- SQL Dashboards: Business reporting.
Interview tip: Explain Databricks as a unified data platform, not just a Spark tool.
Q2. Explain Databricks architecture.
Answer:

Databricks architecture primarily consists of two major components:
1. Control Plane.
2. Compute Plane.
1. Control Plane

The Control Plane manages workspace-level services and orchestration.

Responsibilities:

- Notebook management.
- Job scheduling.
- Cluster configuration.
- User interface.
- Platform services.
2. Compute Plane

The Compute Plane executes data-processing workloads.

Responsibilities:
- Spark job execution.
- Data transformations.
- Distributed processing.
- Task execution.
- Data access.

Q4. What is a Databricks cluster?
Answer:

A Databricks cluster is a collection of compute resources used to execute distributed data-processing workloads using Apache Spark.

It generally consists of:

1. Driver Node
- Coordinates Spark execution.
- Creates the Spark execution plan.
- Schedules tasks.
- Collects task results.
2. Worker Nodes
- Execute distributed tasks.
- Process data partitions.
- Perform transformations.
- Store intermediate computation data.
3. Executors
- Execute tasks assigned by the driver.
- Perform computations.
- Maintain intermediate data and cached partitions.
Real-time example:

Suppose my E-Commerce orders table contains 20 million records. Spark distributes the data-processing work across worker resources. The driver coordinates execution, while executors process different partitions in parallel.

Interview tip: The driver coordinates the job; workers execute distributed tasks. Avoid saying the driver processes every record itself.

SECTION 5: DATABRICKS SECRETS
Q19. What are Databricks Secrets?
Answer:

Databricks Secrets provide a secure mechanism for storing and retrieving sensitive information such as passwords, API keys, access tokens, and database credentials.

Instead of hardcoding credentials inside notebooks, we store them in secret scopes and retrieve them during execution.

Important terminology:
Term	Meaning
Secret Scope	Logical container for secrets
Secret	Sensitive key-value information
Secret Key	Name used to retrieve a secret
ACL	Access Control List for managing permissions


Real-time example:

In an E-Commerce data pipeline, if I need to connect to an external PostgreSQL database, I store the database username and password in Databricks Secrets.
Q20. How do you create a Secret Scope?
Answer:

We can create secret scopes using the Databricks CLI.

Step 1: Configure CLI authentication

Install and authenticate the Databricks CLI using an approved authentication method.

Step 2: Create scope
databricks secrets create-scope ecommerce-scope

Step 3: Add database username
databricks secrets put-secret ecommerce-scope db-username

Step 4: Add database password
databricks secrets put-secret ecommerce-scope db-password

The CLI supports entering the secret values through its interactive input flow.

Step 5: Grant permissions
databricks secrets put-acl ecommerce-scope data-engineering-group READ

Real-time example:

I create an ecommerce-scope and store PostgreSQL credentials inside it. Authorized pipeline users can retrieve those credentials without storing them directly in notebook source code.

Interview tip: Secret scope permissions should follow least-privilege principles.

Q22. How do you use Databricks Secrets in a JDBC connection?
Answer:

JDBC stands for Java Database Connectivity. It allows Spark to connect to external relational databases such as PostgreSQL, MySQL, and SQL Server through supported JDBC drivers.

Q24. What are the types of Databricks Volumes?
Answer:

Databricks supports two types of Unity Catalog Volumes.

1. Managed Volume
- Databricks manages the underlying storage location.
- Unity Catalog manages permissions.
- Suitable for files whose storage location is managed through Databricks.
2. External Volume
- References an existing external cloud storage location.
- Unity Catalog governs access to the files.
- Suitable when organizations already maintain cloud storage locations.
Managed Volume	External Volume
Managed storage location	Existing external storage location
Databricks manages storage placement	References external storage
Unity Catalog permissions	Unity Catalog permissions


Real-time example:

If my project stores incoming CSV files in a Databricks-managed storage location, I can use a Managed Volume. If the organization already has files in an existing cloud storage location, an External Volume may be appropriate.

Q26. What is the difference between DBFS and Unity Catalog Volumes?
Answer:

DBFS stands for Databricks File System. It is a legacy filesystem abstraction used in older Databricks environments.

Unity Catalog Volumes are governed file-storage objects integrated with Unity Catalog.
DBFS	Unity Catalog Volumes
Legacy filesystem abstraction	Unity Catalog-governed file storage
Legacy DBFS root may be disabled	Recommended for governed file access
Legacy paths and access patterns	Supports Unity Catalog permissions
Often found in older projects	Suitable for modern governed architectures


Real-time example:

Interview tip: Do not say DBFS and Volumes are exactly the same. Volumes provide a Unity Catalog governance model for file access.

Q40. What happens when a Databricks cluster terminates?
Answer:

When a Databricks cluster terminates, its compute resources are released and Spark processing stops.

The following things happen:
1. Running Spark jobs stop or fail.
2. In-memory cached DataFrames are lost.
3. Driver and executor processes terminate.
4. Compute resources are released.
5. Workspace notebooks remain available.
6. Persistent Delta tables remain stored in cloud storage.
7. Data stored in Unity Catalog Volumes remains available.
Real-time example:

Suppose I process 10 million E-Commerce order records and save the output into a Silver Delta table. If the cluster terminates after successful execution, the Silver Delta table remains available because it is stored persistently, not in cluster memory.

Interview tip: Cluster termination does not mean persistent data deletion.
