### 1. Project Overview


The E-Commerce Lakehouse is an end-to-end data engineering project developed using Databricks, PySpark, Spark SQL, Delta Lake, and Databricks AI/BI Dashboards.

The primary objective is to build a scalable data processing platform that ingests e-commerce data from multiple business domains, cleans and validates the data, transforms it into business-ready analytical datasets, and presents the results through interactive dashboards.

The project handles seven major datasets:

1. Customers
2. Products
3. Orders
4. Order Items
5. Payments
6. Shipments
7. Inventory Snapshots

The project follows the Medallion Architecture:

- Bronze Layer – Raw data ingestion.
- Silver Layer – Data cleansing and validation.
- Gold Layer – Business transformations and analytics.

It also includes incremental ingestion, data quality validation, referential integrity checks, quarantine handling, notebook orchestration, logging, checkpoint management, monitoring, end-to-end testing, and project readiness reporting.


### 1.1 Business Problem



E-commerce organizations generate large amounts of data across different systems.

Customer information, product details, orders, payments, shipments, and inventory records are stored separately. These datasets may contain:

- Missing values.
- Duplicate records.
- Invalid data types.
- Incorrect date formats.
- Inconsistent business statuses.
- Invalid foreign key relationships.
- Incomplete payment information.
- Incorrect shipment statuses.
- Inventory inconsistencies.

Processing these datasets manually is difficult and makes business reporting unreliable.

The solution is to develop a centralized Lakehouse platform that processes raw data, validates it, creates trusted analytical datasets, and provides business insights.

### Auto Loader


Auto Loader is a **Databricks feature** used to `automatically load new files from cloud storage` into `Delta Lake` using Spark Structured Streaming.



**Important configurations:**

- `cloudFiles.format`
- `cloudFiles.schemaLocation`
- `checkpointLocation`
- `header`
- `trigger(availableNow=True)`


### Conceptual example:


```python
customers_df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option(
        "cloudFiles.schemaLocation",
        schema_path
    )
    .load(customers_landing_path)
)

(
    customers_df.writeStream
    .format("delta")
    .option(
        "checkpointLocation",
        checkpoint_path
    )
    .trigger(availableNow=True)
    .toTable(
        "workspace.ecommerce_bronze.bronze_customers"
    )
)
```


- This is an illustrative Auto Loader pattern. 
- The exact schema, paths, options, and sink behavior must follow the actual notebook implementation.


## Why Auto Loader?

- Supports incremental file discovery.
- Tracks processed files using checkpoints.
- Supports schema tracking and evolution options.
- Reduces the need to manually identify newly arriving files.

### Null Validation



Identify records containing `missing mandatory values`.


**Example:**


```python
from pyspark.sql import functions as F

null_records = customers_df.filter(
    F.col("customer_id").isNull()
)
```


### Duplicate Detection


Identify `duplicate business keys`.


```python
duplicate_records = (
    customers_df
    .groupBy("customer_id")
    .count()
    .filter(F.col("count") > 1)
)
```


### Data Type Validation


Convert `columns` into the `expected data types`.


```python
orders_df = (
    orders_df
    .withColumn(
        "order_total",
        F.col("order_total").cast("decimal(18,2)")
    )
    .withColumn(
        "order_date",
        F.to_date("order_date")
    )
)
```


### Business Rule Validation


**Example rules:**


- Order amount should not be negative.
- Quantity should be greater than zero.
- Required identifiers should not be null.
- Payment amounts should follow the defined business rules.
- Shipment statuses should be standardized.

## Example Gold Transformation

Daily sales aggregation:

```python
from pyspark.sql import functions as F

daily_sales_df = (
    orders_df
    .filter(
        F.col("order_status") != "CANCELLED"
    )
    .groupBy(
        F.to_date("order_date").alias("order_date"),
        "currency"
    )
    .agg(
        F.countDistinct("order_id").alias(
            "total_orders"
        ),
        F.sum("order_total").alias(
            "gross_order_value"
        )
    )
)
```

This example demonstrates a possible daily sales aggregation. The exact revenue definition and filtering logic must match the implemented Gold notebook.

## 19. Frequently Asked Interview Questions


### Q1. Why did you use Databricks?

**Databricks** provides a managed environment for `distributed Spark processing, SQL analytics, Delta Lake, notebook development, governance,` and `workflow orchestration`.


### Q2. What is Medallion Architecture?

- It is a **data layered architecture** consisting of `Bronze, Silver,` and `Gold layers`. 
- Bronze stores `raw data`, 
- Silver contains `cleaned data`, 
- Gold contains `business-ready analytical datasets`.


### Q3. Why did you use Auto Loader?

- **Auto Loader** is a `Databricks feature` used to `automatically load new files from cloud storage` into `Delta Lake` using Spark Structured Streaming.
- It is used for `file-based incremental ingestion`.


### Q4. What is Delta Lake?

- **Delta Lake** is a `storage layer` that provides `ACID transactions, schema enforcement, table history,` and 
- `operations` such as `MERGE, UPDATE, and DELETE`.


### Q5. What is a checkpoint?

- **Checkpointing** is a `fault-tolerance mechanism` used to stores `streaming query progress` and `state information` in storage.
- If a failure occurs, Spark can `recover and continue processing` using the **checkpointing**.


### Q6. What is a shuffle?

- **Shuffle** is the process of` redistributing data across different partitions or executors` in Spark.
- It is commonly occurs in `wide transformations`.
- Common Operations of shuffle are `groupBy(), distinct(), orderBy(), repartition(), Sort-based joins`.


### Q7. What is data skew?

- **Data skew** occurs when `data is distributed unevenly across partitions`.
- **Data Skew** is handle by `Salting, Broadcast Join, Adaptive Query Execution (AQE), Repartitioning, Filtering`.

**Example:**

Suppose we have 1 million employee records distributed across 4 partitions.

**Balanced data distribution:**

Partition 1 - 250K records

Partition 2 - 250K records

Partition 3 - 250K records

Partition 4 - 250K records

**Skewed data distribution:**

Partition 1 - 100K records

Partition 2 - 100K records

Partition 3 - 100K records

Partition 4 - 700K records


### Q8. What is the difference between repartition and coalesce?

**Repartition:**

- **repartition()** is used to `increases or decreases a partitions` by `redistributing data evenly across partitions`.
- Performs a `shuffle`.
- `More expensive`.
- Better for `redistributing data evenly`.

**Coalesce:**

- **coalesce()** is used to `decreases partitions`.
- Usually avoids a `full shuffle`.
- `Less expensive`.
- Better for `reducing the number of partitions`.


### Q9. What is referential integrity?

A **referential integrity** ensures that **foreign key values** in a `child table` correspond to **valid primary key values** in a `parent table`.


### Q10. How did you handle duplicate records?

- I identified `business keys` and `used duplicate detection` and `deduplication` logic to identify **repeated records**. 
- For deterministic deduplication, a window function with an `appropriate timestamp` and `tie-breaker` can be used.


### Q11. How did you handle invalid records?

- I applied `validation rules` and 
- `separated the failed records with required checks into quarantine tables` for **further investigation.**



### Q12. How did you orchestrate the pipeline?

- I created a **parent notebook** that invokes `individual notebooks sequentially` using `dbutils.notebook.run()`, `passes parameters`, `captures execution results`. 
- stops the `workflow` when a **required task fails**.


### Q13. How did you monitor pipeline execution?

- I maintained **execution logs** that containing `run IDs, task names, statuses, start and end timestamps, durations,` and `error messages`.
- A **monitoring notebook** analyzes those `records and generates task` and `pipeline health reports`.


### Q14. How did you validate Gold data?

I used 

- source-to-target reconciliation, 
- record counts, 
- business rule checks, 
- relationship validation, 
- aggregation checks


### Q15. What would you improve?

I would improve

- selective Gold refresh,

- controlled recovery testing, 
- scheduled workflow execution,  
- automated alerting
