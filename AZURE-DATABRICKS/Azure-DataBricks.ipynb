
## 1. Azure Databricks Fundamentals


### Q1. What is Azure Databricks?


- **Azure Databrick**s is a `cloud-based data analytics platform` built around **Apache Spark**. 
- It is used for `big data processing, transformation, analytics` and `machine learning`.


| Component | Purpose |
|---|---|
| Workspace | Develop and manage notebooks, queries and jobs |
| Notebooks | Write and execute PySpark, Python and SQL code |
| Apache Spark | Distributed data processing engine |
| Delta Lake | Provides reliable data storage with ACID transactions |
| Unity Catalog | Centralized data governance and access control |
| Lakeflow Jobs | Schedule and orchestrate data processing tasks |
| SQL Warehouse | Execute analytical SQL queries |


### Q2. What is the architecture of Azure Databricks?


**Azure Databricks architecture** mainly consists of `two planes`.


| Component | Explanation |
|---|---|
| **Control Plane** | Manages the Databricks workspace, notebooks, jobs, cluster configuration, user access and other management services. |
| **Compute Plane** | Executes Spark workloads, runs queries and performs data processing using compute resources. |


**Other important components:**

- Workspace
- Clusters / Serverless Compute
- Notebooks
- Jobs and Workflows
- Unity Catalog
- Delta Lake
- ADLS Gen2

2. **Job compute** – Used for executing scheduled or automated jobs.

### Q13. What is Apache Spark?


**Apache Spark** is an `open-source distributed computing framework` used to `process large datasets across multiple machines`.


**It supports:**

- Batch processing.
- Stream processing.
- SQL processing.
- Machine learning.

### Q14. What is the difference between Transformation and Action in Spark?


| Transformation | Action |
|---|---|
| Creates a new DataFrame or RDD | Triggers execution of the computation |
| Usually lazy in execution | Executes the required computation |
| Does not immediately return final results | Returns results or writes data |
| Examples: `filter()`, `select()`, `withColumn()` | Examples: `count()`, `collect()`, `show()` |


### Q15. What is Lazy Evaluation in Spark?


Lazy evaluation means Spark does not execute transformations immediately.

It builds a logical execution plan and executes the required operations when an action is called.

Example:
df1 = df.filter("salary > 30000")df2 = df1.select("name", "salary")df2.show()



Spark executes the required computation when show() is called.


### Q16. What is the difference between Narrow and Wide Transformations?


Narrow Transformation	Wide Transformation
No shuffle between partitions is required	Requires data exchange between partitions
Generally faster	Generally more expensive
Data can be processed locally	Data may move across executors
filter(), map()	groupBy(), distinct(), repartition()


### Q17. What is Shuffle in Spark?


Shuffle is the process of redistributing data across partitions or executors.

It usually happens during operations such as:
- groupBy()
- join()
- distinct()
- orderBy()
- repartition()
Shuffle can increase execution time because it involves network communication and disk I/O.


### Q18. What is the difference between repartition() and coalesce()?



repartition()	coalesce()
Can increase or decrease partitions	Primarily decreases partitions
Usually involves shuffle	Usually avoids full shuffle
More expensive	Generally less expensive
Used for redistributing data	Used to reduce partitions

Example:
df.repartition(10)df.coalesce(2)


### Q19. What is a Broadcast Join?


A Broadcast Join is used when one DataFrame is small enough to be distributed to all Spark executors.

It avoids shuffling the large DataFrame for that join.

Example:
from pyspark.sql.functions import broadcastresult = orders.join(    broadcast(customers),    "customer_id",    "inner")



Real-time example:

Joining a 10 GB orders table with a 5 MB customer lookup table.


### Q20. What is Data Skewness in Spark?


Data skewness occurs when data is distributed unevenly across partitions.

For example:

A customer with 10 million orders while other customers have only a few hundred orders.

Solutions:
- Salting.
- Broadcast joins where appropriate.
- Repartitioning using suitable keys.
- Adaptive Query Execution (AQE).
- Filtering unnecessary data.


### Q21. What is the difference between RDD, DataFrame and Dataset?


RDD	DataFrame	Dataset
Low-level abstraction	Higher-level abstraction	Typed abstraction
Less optimization	Catalyst optimization	Catalyst optimization
More manual operations	SQL-like operations	Strongly typed operations
Available in PySpark	Available in PySpark	Not available as a typed API in PySpark


### Q22. What is Catalyst Optimizer?


Catalyst is Spark SQL's query optimization framework.

It optimizes logical and physical query plans to improve execution performance.

Examples:
- Predicate pushdown.
- Constant folding.
- Projection pruning.
- Join optimization.

### Q23. What is Delta Lake?


Delta Lake is an open-source storage framework that provides reliability and transactional capabilities on top of data lake storage.

**It supports:**
- ACID transactions.
- Schema enforcement.
- Schema evolution.
- Time Travel.
- MERGE, UPDATE and DELETE.
- Change Data Feed.
- Transaction log.


### Q24. What is the difference between Data Lake, Delta Lake and Lakehouse?


Data Lake	Delta Lake	Lakehouse
Stores raw data	Adds reliability to data lake storage	Combines lake flexibility with warehouse capabilities
Supports multiple file formats	Uses Parquet and transaction logs	Uses technologies such as Delta Lake
Limited transactional capabilities	Supports ACID transactions	Supports analytics and data engineering
Usually object storage	Storage framework	Overall architecture


### Q25. What are ACID properties in Delta Lake?


ACID represents four transaction properties.
1. Atomicity: A transaction either completes fully or fails fully.
2. Consistency: Every successful transaction maintains valid table state.
3. Isolation: Concurrent transactions do not expose partially committed changes.
4. Durability: Once committed, changes are persisted.


### Q26. What is the Delta Transaction Log?


The Delta Transaction Log is a collection of JSON commit files and checkpoint Parquet files stored inside the _delta_log directory.

It maintains information about:
- Table transactions.
- Added files.
- Removed files.
- Schema changes.
- Table metadata.
- Protocol versions.
It helps Delta Lake determine the current version and state of a table.


### Q27. What is Time Travel in Delta Lake?


Time Travel allows us to access previous versions of a Delta table.

Example:
df = spark.read.format("delta") \    .option("versionAsOf", 2) \    .load("/path/to/delta_table")display(df)



We can also query using SQL:
SELECT *
FROM employees VERSION AS OF 2;


### Q28. What is VACUUM in Delta Lake?


VACUUM removes obsolete data files that are no longer required by the retained Delta table history.

Example:
VACUUM employees RETAIN 168 HOURS;


Important: VACUUM does not simply delete old transaction log history. Log cleanup and data-file cleanup are separate processes.


### Q29. What is Schema Enforcement?


Schema enforcement ensures incoming data matches the existing Delta table schema.

If incompatible data is inserted, the operation may fail.

Example:

An existing table has:
employee_id INT
salary DOUBLE


Incoming data has an incompatible salary type. Schema enforcement can reject the write.


### Q30. What is Schema Evolution?


Schema evolution allows supported schema changes during data writes.

Example:

Existing schema:
employee_id
employee_name


New incoming schema:
employee_id
employee_name
department


Example:
df.write.format("delta") \    .option("mergeSchema", "true") \    .mode("append") \    .save("/path/to/table")



Schema migration and schema evolution are related, but they are not the same.



Q31. What is MERGE in Delta Lake?
MERGE is used to perform INSERT, UPDATE and DELETE operations based on matching conditions.

Example:
MERGE INTO target_employees AS target
USING source_employees AS source
ON target.employee_id = source.employee_id

WHEN MATCHED THEN
  UPDATE SET *

WHEN NOT MATCHED THEN
  INSERT *;


Real-time example:

If an employee already exists, update their salary. If the employee is new, insert a new record.


### Q32. What is Change Data Feed (CDF)?


**Change Data Feed** tracks row-level changes made to a Delta table.

**It can capture:**
- Inserted records.
- Updated records.
- Deleted records.

It is useful for incremental data processing and downstream synchronization.

### Q36. What is Incremental Data Loading?


Incremental loading means processing only new or modified records instead of processing the entire dataset every time.

Common approaches:
- Watermark-based loading.
- Last modified timestamp.
- CDC.
- Delta Change Data Feed.
- Auto Loader.
Example:
from pyspark.sql.functions import colnew_df = source_df.filter(    col("updated_at") > last_processed_timestamp)

**It supports:**

- Incremental ingestion.

### Q46. How do you optimize Spark performance in Databricks?


**Important optimization techniques:**

1. Use appropriate partitioning.
2. Avoid unnecessary shuffles.
3. Use broadcast joins for suitable small tables.
4. Avoid unnecessary collect().
5. Use Parquet or Delta instead of CSV for analytical workloads.
6. Apply predicate pushdown.
7. Use Adaptive Query Execution.
8. Avoid excessive small files.
9. Use appropriate cluster sizing.
10. Use Liquid Clustering or suitable data layout techniques.


### Q47. What is Photon in Databricks?


Photon is Databricks' vectorized query execution engine designed to accelerate SQL and DataFrame workloads.

It can improve performance for supported operations by using optimized execution techniques.
Q48. What is Adaptive Query Execution (AQE)?
AQE is a Spark SQL optimization feature that adjusts the execution plan at runtime using actual data statistics.

It can:
- Optimize join strategies.
- Coalesce shuffle partitions.
- Handle certain skewed joins.


### Q49. What is Partitioning in Spark?


Partitioning divides a dataset into smaller logical pieces so Spark can process them in parallel.

Example:
df.write.format("delta") \    .partitionBy("department") \    .save("/path/to/employees")



Partitioning can improve query performance when queries filter on suitable partition columns. However, excessive partitioning can create many small files.


### Q50. What is the difference between OPTIMIZE and ZORDER?


OPTIMIZE	ZORDER

Note: Liquid Clustering is the recommended modern layout approach for many new Delta table designs; ZORDER is still relevant for existing workloads and supported configurations.
